```python
import json
import os
import pandas as pd

target_json_path = './data/extracted_saramin_jobs_result.json'  # 내 파일
new_json_path = './data/extracted_jobs_result_saramin.json' # 민석오빠 파일
final_path = './data/final_extracted_result.json'
```

```python
existing_data = [] # 내 파일 데이터 저장
if os.path.exists(target_json_path):
    try:
        with open(target_json_path, 'r', encoding='utf-8') as f:
            existing_data = json.load(f)
        print(f"🔄 기존 메인 파일에서 {len(existing_data):,}건의 데이터를 읽어왔습니다.")
    except Exception as e:
        print(f"⚠️ 기존 메인 파일 읽기 실패: {e}")

new_data_to_append = [] # 민석오빠 파일 데이터 저장
if os.path.exists(new_json_path):
    try:
        with open(new_json_path, 'r', encoding='utf-8') as f:
            new_data_to_append = json.load(f)
        print(f"📥 새 JSON 파일에서 {len(new_data_to_append):,}건의 데이터를 읽어왔습니다.")
    except Exception as e:
        print(f"❌ 새 JSON 파일 읽기 실패: {e}")
        exit() # 새 파일을 못 읽으면 중단
else:
    print(f"❌ 새로 붙일 JSON 파일({new_json_path})이 존재하지 않습니다.")
    exit()

# 데이터 리스트 하나로 합치기
all_data = existing_data + new_data_to_append

# 중복 제거 및 원문 null 데이터 제외 
final_cleaned_data = []
processed_ids = set()
removed_null_count = 0
removed_dup_count = 0

for item in all_data:
    job_id = item.get('공고번호')
    raw_text = item.get('원문')
    
    # 원문이 null, 빈 문자열, NaN이면 제외
    if raw_text is None or str(raw_text).strip() == "" or pd.isna(raw_text):
        removed_null_count += 1
        continue
        
    # 중복 제거
    if job_id in processed_ids:
        removed_dup_count += 1
        continue
        
    # 두 조건을 모두 통과한 데이터만 저장 목록에 추가
    processed_ids.add(job_id)
    final_cleaned_data.append(item)

# 최종 데이터를 기존 메인 파일에 덮어쓰기
with open(final_path, 'w', encoding='utf-8') as f:
    json.dump(final_cleaned_data, f, ensure_ascii=False, indent=4)
```
